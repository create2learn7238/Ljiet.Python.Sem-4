```python
import requests
from bs4 import BeautifulSoup

soup=BeautifulSoup(open("top250.html","r",encoding="utf-8"))
# print(soup.prettify())
```

```python
y=soup.find_all("td" ,class_="titleColumn")
title=[]
for i in y:
    title.append(i.get_text(strip=True).split(".")[1].split("(")[0])
```

```python
y=soup.find_all("td" ,class_="ratingColumn imdbRating")
rate=[]
for i in y:
    rate.append(i.get_text(strip=True))
```

```python
y=soup.find_all("span" ,class_="secondaryInfo")
year=[]
for i in y:
    y=i.text[1:-1]
    year.append(y)
```

```python
import pandas as pd
df=pd.DataFrame({"title":title,"year":year,"rate":rate,})
df
```
