# https://www.politifact.com/factchecks/

```python
import requests
from bs4 import BeautifulSoup
url="https://www.politifact.com/factchecks/"
result=requests.get(url)
content=result.text
soup=BeautifulSoup(content , "html.parser")
```

```python
x=soup.find_all("a",href=True,class_="m-statement__name")
source=[]
for a in x:
    source.append(a.get_text(strip=True))
```

```python
x=soup.find_all("footer",class_="m-statement__footer")
date=[]
for a in x:
    date.append(a.get_text(strip=True).split("•")[1])
```

```python
x=soup.find_all("div",class_="m-statement__quote")
quote=[]
for a in x:
    quote.append(a.get_text(strip=True))
```

```python
x=soup.find_all("footer",class_="m-statement__footer")
owner=[]
for a in x:
    owner.append(a.get_text(strip=True).split("•")[0])
owner=[name.replace('By ', '').strip() for name in owner]
```

```python
import pandas as pd
df=pd.DataFrame({"Owner":owner,"Date":date,"Quote":quote,"Source":source})
df
```
