```python
import requests
from bs4 import BeautifulSoup

soup=BeautifulSoup(open("Popular_Quotes.html","r",encoding="utf-8"))
# print(soup.prettify())
```

```python
quoteText=[]
author=[]
for i in soup.find_all("div" ,class_="quoteText"):
    quoteText.append(i.get_text(strip=True).split("―")[0])
    author.append(i.get_text(strip=True).split("―")[1])
```

```python
like=[]
for i in soup.find_all("a" ,class_="smallText"):
    like.append(i.get_text(strip=True).split("likes")[0])
```

```python
import pandas as pd
df=pd.DataFrame({"quote":quoteText,"author":author,"like":like})
df
```

```python
df.sort_values(by="like")
```
