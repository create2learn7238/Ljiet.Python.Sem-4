# Call 5 day / 3 hour forecast data

```python
import requests,json
from bs4 import BeautifulSoup
api="3b5a258fb2bb246eacb62e89cca5fa33"
def forecast():
    api_key=api
    url=f"https://api.openweathermap.org/data/2.5/forecast?lat={x}&lon={y}&appid={api}"
    response=requests.get(url,verify=False)
    forecast=response.json()
    print(json.dumps(forecast,indent=5))
    return forecast
a=forecast()
```

```python
date=[]
desc=[]
temp=[]
temp=[]
pressure=[]
humidity=[]
visibility=[]
windspeed=[]
for i in a["list"]:
    date.append(i["dt_txt"])
    desc.append(i["weather"][0]["description"])
    celsius_temp = i["main"]["temp"] - 273.15
    temp.append(round(celsius_temp, 2))    
    pressure.append(i["main"]["pressure"])
    humidity.append(i["main"]["humidity"])
    visibility.append(i["visibility"])
    windspeed.append(i["wind"]["speed"])
import pandas as pd

# 1. Create a dictionary pairing column names with your lists
weather_data = {
    "Date": date,
    "Description": desc,
    "Temperature (°C)": temp,
    "Pressure (hPa)": pressure,
    "Humidity (%)": humidity,
    "Visibility (m)": visibility,
    "Wind Speed (m/s)": windspeed
}

# 2. Create the DataFrame
df = pd.DataFrame(weather_data)
df
```
