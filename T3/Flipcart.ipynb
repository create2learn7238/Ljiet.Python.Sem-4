```python
import requests
from bs4 import BeautifulSoup

soup=BeautifulSoup(open("flipkart.html","r",encoding="utf-8"))
# print(soup.prettify())
```

```python
name=[]
rom=[]
color=[]
for i in soup.find_all("div" ,class_="KzDlHZ"):
    name.append(i.get_text(strip=True).split("(")[0].strip())
    color.append(i.get_text(strip=True).split("(")[1].split(",")[0].strip())
    rom.append(i.get_text(strip=True).split("(")[1].split(",")[1].strip()[0:-1])
```

```python
price=[]
for i in soup.find_all("div" ,class_="Nx9bqj _4b5DiR"):
    price.append(i.get_text(strip=True))
```

```python
rate=[]
for i in soup.find_all("div" ,class_="XQDdHH"):
    rate.append(i.get_text(strip=True))
rate=rate[0:24]
```

```python
ram=[]
for i in soup.find_all("ul" ,class_="G4BRas"):
    ram.append(i.get_text(strip=True).split("RAM")[0].strip()+" RAM")
```

```python
import pandas as pd
df=pd.DataFrame({"name":name,"rom":rom,"color":color,"price":price,"rate":rate,"ram":ram})
df
```
